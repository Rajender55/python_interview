```python
#prime factorization 
def is_prime(num):
    if num<=1:
        return False
    else:
        for i in range(2,num):
            if num%i==0:
                break
                return False
        else:
            return True
def prime_factorization(num):
    for i in range(2,num+1):
        if num%i==0:
            if is_prime(i):
                print(i)
                num=num//i
                return prime_factorization(num)

prime_factorization(7)
```
