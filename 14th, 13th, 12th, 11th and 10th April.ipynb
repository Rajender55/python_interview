 **1.Sum of even elements in a given list**

```python
l=list(map(int, input().split()))

def sum_of_even(l: list) -> list:
    return sum([x for x in l if x%2==0])

print(f'sum of the even elements in {l} is: ',sum_of_even(l))
```

**2.Unique elements in a given list**

```python
l=list(map(int, input().split()))

def unique_elements(l:list)->list:
    return list(set(l))

uni=unique_elements(l)

print(f'Unique elements in {l} are: ',uni)
```

**3.intersected elements in 2 elements**

```python
l1=list(map(int, input().split()))
l2=list(map(int, input().split()))

def common_el(l1:list,l2:list)->list:
    return list(set(l1).intersection(set(l2)))

print(common_el(l1,l2))
```

**4.product of numbers in the list**

```python
l=list(map(int, input().split()))

def product_of_numbers(l:list)->int:
    fact=1
    for i in l:
        fact=fact*i
    return fact

print(f'The product of numbers in {l}: {product_of_numbers(l)}')
```

**11.Return the elements that are only in one of the lists**

```python
l1=list(map(int, input().split()))
l2=list(map(int, input().split()))

def numbers_once(l1:list, l2:list)->list:
    return list(set(l1).symmetric_difference(set(l2)))

print(f'Elements that are present in only once are: {numbers_once(l1,l2)}')
```
