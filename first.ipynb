```python
#2. if a number if prime or not and n<100 print all
import math
n=100
prime = []
for i in range(2, n+1):
    for j in range(2, int(math.sqrt(i))+1):
        if i%j==0:
            break
    else:
        prime.append(i)
print(prime)
```

```python
#3.febinocci sequence up to n terms
n=7
feb=[]
if n==1:
    feb.append(1)
elif n==2:
    feb.extend([1,1])
else:
    feb.extend([1, 1])
    a=1; b=1; i=2
    while i<n:
        c=a+b 
        feb.append(c)
        a=b 
        b=c 
        i+=1
print(feb)
```

```python
#4.function to find factorial of a number using recursion
def fact_num(x):
    if x == 1 or x==0:
        return 1
    else:
        return x*fact_num(x-1)
    
```

```python
ou = fact_num(7)
print(ou)
```

```python
#method 2
def rev_string(s):
    if len(s)==0:
        return s
    else:
        return rev_string(s[1:])+s[0]
k='hello'
print(rev_string(k))
```

```python
#6.Write a program to check if a string is a palindrome.
k='baby'
if k==k[::-1]:
    print('Its a palindrome')
else:
    print("Not a palindrome")

```

```python
#method 2
def is_palidrome(s, i, j):
    if i>=j:
        return True
    if s[i]!=s[j]:
        return False
    else:
        return is_palidrome(s, i+1, j-1)
k = 'bab'
o = is_palidrome(k, 0, len(k)-1)
print(o)
```

```python
#7.Write a function to count distinct vowels and consonants in a string.
def count(s):
    v=[]
    c=[]
    for i in range(len(s)):
        if s[i].isalpha():
            if s[i] in ['a', 'e', 'i', 'o', 'u', 'A', 'E', 'I', 'O', 'U']:
                if s[i] not in v:
                    v.append(s[i])
            else:
                if s[i] not in c:
                    c.append(s[i])
    return len(v),len(c) 
k=count('rajen234der')
print(k)
```
